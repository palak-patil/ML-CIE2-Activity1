## Step 3: Identifying Data Types and Basic Statistics

The columns in the Titanic dataset are classified as numerical, categorical or Boolean based on their data types.

Numerical columns contain numbers, while categorical columns represent groups or labels. Summary statistics help us understand the distribution of numerical data.

```python
# Identify numerical columns
numerical_columns = df.select_dtypes(include="number").columns.tolist()

# Identify categorical columns
categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Numerical columns:")
print(numerical_columns)

print("\nCategorical columns:")
print(categorical_columns)
```

## Step 4: Identifying Missing Values

Missing values occur when information is not available for certain records. They may affect data analysis and machine learning results.

In this step, we identify missing values in each column and calculate their percentage to understand which columns need attention.

```python
# Count missing values in each column
missing_count = df.isnull().sum()

# Calculate the percentage of missing values
missing_percentage = (missing_count / len(df)) * 100

# Create a summary table
missing_summary = pd.DataFrame({
    "Missing Values": missing_count,
    "Percentage (%)": missing_percentage
})

# Display columns with missing values first
missing_summary.sort_values(
    by="Missing Values",
    ascending=False
)
```

### Observation

The missing-value analysis shows that the dataset contains missing information in some columns. The `deck` column has a large number of missing values, while `age`, `embarked` and `embark_town` contain fewer missing values.

These missing values need to be handled before the dataset is used for further analysis and machine learning.

## Step 5: Handling Missing Values

A copy of the original dataset is created to preserve the original data. Missing numerical values are filled using the median, while missing categorical values are filled using the mode. The `deck` column is removed because it contains a large number of missing values.

```python
# Create a copy of the original dataset
df_clean = df.copy()

# Remove the deck column due to excessive missing values
df_clean = df_clean.drop(columns=["deck"])

# Fill missing numerical values using the median
df_clean["age"] = df_clean["age"].fillna(df_clean["age"].median())

# Fill missing categorical values using the mode
df_clean["embarked"] = df_clean["embarked"].fillna(
    df_clean["embarked"].mode()[0]
)

df_clean["embark_town"] = df_clean["embark_town"].fillna(
    df_clean["embark_town"].mode()[0]
)

# Check missing values after handling
df_clean.isnull().sum()
```

## Step 6: Checking Duplicates and Data Consistency

Duplicate records and inconsistent category labels can affect data quality. In this step, duplicate rows are identified and removed, and category labels are checked for unnecessary spaces or inconsistent formats.

```python
# Count duplicate rows
print("Duplicate rows:", df_clean.duplicated().sum())

# Remove duplicate rows
df_clean = df_clean.drop_duplicates()

# Remove extra spaces from text columns
for col in df_clean.select_dtypes(include="object").columns:
    df_clean[col] = df_clean[col].str.strip()

# Check duplicates after cleaning
print("Remaining duplicates:", df_clean.duplicated().sum())
```

## Step 7: Exploratory Data Analysis (EDA)

Exploratory Data Analysis (EDA) helps us understand the distribution and patterns in a dataset using statistics and visualizations. In this step, we analyse passenger ages, survival status and passenger class.

```python
# Summary statistics for numerical columns
df_clean.describe()
```

```python
# Count passengers in each passenger class
df_clean["class"].value_counts()
```

```python
# Histogram showing the distribution of passenger ages
plt.figure(figsize=(8, 5))
sns.histplot(df_clean["age"], bins=20, kde=True)

plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
# Bar chart showing survival by passenger class
sns.countplot(data=df_clean, x="class", hue="survived")

plt.title("Passenger Survival by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.legend(title="Survived", labels=["No", "Yes"])
plt.show()
```

### Observations

2. The number of passengers who did not survive was higher in Third Class than in First Class.
3. The graphs show differences in passenger age distribution and survival across passenger classes.

## Step 8: Encoding Categorical Features

One-hot encoding converts categorical values into separate numerical columns containing 0s and 1s. It is suitable for nominal categories that do not have a natural order.

In this step, the `sex`, `embarked`, `class` and `who` columns are encoded using one-hot encoding.

```python
# Create a copy of the cleaned dataset
df_encoded = df_clean.copy()

# One-hot encode selected categorical columns
df_encoded = pd.get_dummies(
    df_encoded,
    columns=["sex", "embarked", "class", "who"],
    dtype=int
)

# Display the first five rows
df_encoded.head()
```

```python
# Display the new column names
print(df_encoded.columns.tolist())
```

### Observation

One-hot encoding converted the selected categorical columns into numerical columns containing 0s and 1s. This allows machine learning algorithms to process categorical information without assigning an artificial numerical order to the categories.

## Step 9: Scaling Numerical Features

Standardisation transforms numerical features so that they have a mean of approximately 0 and a standard deviation of approximately 1.

It is useful for machine learning algorithms that are sensitive to differences in feature scales. Identifier columns and target variables are not scaled.

```python
from sklearn.preprocessing import StandardScaler

# Create a copy of the encoded dataset
df_prepared = df_encoded.copy()

# Select numerical input features for scaling
scale_columns = ["age", "fare", "sibsp", "parch"]

# Apply standardisation
scaler = StandardScaler()
df_prepared[scale_columns] = scaler.fit_transform(
    df_prepared[scale_columns]
)

# Display the prepared data
df_prepared.head()
```

```python
# Check the mean and standard deviation after scaling
print("Mean:")
print(df_prepared[scale_columns].mean())

print("\nStandard deviation:")
print(df_prepared[scale_columns].std())
```

## Step 10: Saving the Prepared Dataset

The prepared dataset is checked for remaining missing values and duplicate records. It is then saved as a separate CSV file, preserving the original dataset.

The final dataset contains cleaned, encoded and scaled features suitable for future machine learning tasks.

```python
# Check the final dataset shape
print("Final dataset shape:", df_prepared.shape)

# Check missing values
print("Total missing values:", df_prepared.isnull().sum().sum())

# Check duplicate rows
print("Total duplicate rows:", df_prepared.duplicated().sum())
```

```python
# Save the prepared dataset
df_prepared.to_csv(
    "prepared_dataset.csv",
    index=False
)

print("Dataset saved successfully!")
```
